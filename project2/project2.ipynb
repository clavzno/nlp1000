````python
data_folder = Path("p2data")
tokenizer = TweetTokenizer(r'\w+')

for lang in languages:
    test = ""
    source_file_path = data_folder / f"(sentences)-{lang}-cleaned.txt"
    output_file_path = data_folder/f"(common-words)-{lang}.txt"

    # if it doesn't exist
    if not source_file_path.exists() or source_file_path.stat().st_size == 0:
        print(f"Skipped: {source_file_path} (file does not exist or is empty)")
        continue

    # rewrite all the txts if it exists
    if output_file_path.exists():
        print(f"unlinking/deleting old files")
        output_file_path.unlink()
    
    # read input
    with source_file_path.open("r", errors="ignore", encoding="utf-8") as f:
        raw_sentences = f.read()
    
    # do whatever you want here
    allWords = tokenizer.tokenize(raw_sentences)

    stopwords = ('.', ',', ';', '?', ':', '?', '(', ')', '"', '\'')
    allWordsFreq = nltk.FreqDist(w.lower() for w in allWords if w not in stopwords)
    for word, count in allWordsFreq.most_common(500):
        strCount = str(count)
        test = test + word+ ":" + strCount + "\n" 

    with output_file_path.open("w", encoding="utf-8") as f:
        f.write(test)

    print(f"Cleaned and saved: {output_file_path}")

````

### 1.2 Using nltk to generate n-grams

We'll be using nltk to generate the n-grams of all the languages we have. We'll save them into files per language.

[Source for words](https://www.pythontutorials.net/nltk/how-to-create-ngrams-with-nltk/)


[Source for characters](https://groups.google.com/g/nltk-users/c/9_dVwH4SBmc)

````python
# sample code that turns bikol sentences into unigrams, bigrams, and trigrams
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

data_folder = Path("p2data")

for lang in languages:
    # define source and output
    source_file_path = data_folder / f"(sentences)-{lang}-cleaned.txt"
    output_file_path = data_folder/f"(trigrams)-{lang}.txt"

    # if source doesn't exist skip
    if not source_file_path.exists() or source_file_path.stat().st_size == 0:
            print(f"Skipped: {source_file_path} (file does not exist or is empty)")
            # continue
    else:       
        # rewrite all the txts if it exists
        if output_file_path.exists():
            print(f"unlinking/deleting old files")
            output_file_path.unlink()

        # read input
        with source_file_path.open("r", errors="ignore", encoding="utf-8") as f:
            raw_sentences = f.read()

        # DO WHATEVER YOU WANT HERE
        # tokens = word_tokenize(raw_sentences) # split into words - DON'T DELETE THIS THANKS
        characters = [c for c in raw_sentences]

        # if you want to do it in words replace "characters" with "tokens"
        # unigrams = list(ngrams(characters, 1))
        # bigrams = list(ngrams(characters, 2))
        trigrams = list(ngrams(characters, 3))

        # save them to the file
        with output_file_path.open("w", encoding="utf-8") as f:
            '''f.write("Unigrams\n")
            for list_tuple in unigrams:
                f.write(" ".join(map(str, list_tuple)) + "\n") 
            f.write("Bigrams\n")
            for list_tuple in bigrams:
                f.write(" ".join(map(str, list_tuple)) + "\n")'''
            f.write("Trigrams\n")
            for list_tuple in trigrams:
                f.write(" ".join(map(str, list_tuple)) + "\n")

        print(f"Cleaned and saved: {output_file_path}")
````

Below, we'll be getting the frequency count of each character trigram per language. Then using the frequency count, we'll write the top 250 trigrams into another file.

````python
from collections import Counter
data_folder = Path("p2data")

for lang in languages:
    # define source and output
    source_file_path = data_folder / f"(trigrams)-{lang}.txt"
    output_file_path = data_folder/f"(top250trigrams)-{lang}.txt"

    # if source doesn't exist skip
    if not source_file_path.exists() or source_file_path.stat().st_size == 0:
            print(f"Skipped: {source_file_path} (file does not exist or is empty)")
            # continue
    else:       
        # rewrite all the txts if it exists
        if output_file_path.exists():
            print(f"unlinking/deleting old files")
            output_file_path.unlink()

        # read input
        with source_file_path.open("r", errors="ignore", encoding="utf-8") as f:
            next(f, None) # skip the first line bc it's just a title
            trigrams = [line.strip() for line in f if line.strip()]

        # DO WHATEVER YOU WANT HERE
        freq_counter = Counter(trigrams)
        top_250 = freq_counter.most_common(250)

        # save them to the file
        with output_file_path.open("w", encoding="utf-8") as f:
            f.write("Top 250\tFrequency\n")
            for trigram, count in top_250:
                 f.write(f"{trigram}\t\t{count}\n")

        print(f"Cleaned and saved: {output_file_path}")
````
